```python
df = pd.read_csv("players.csv")
df_country_colors = pd.read_csv("country_colors.csv")
# Add color columns from df_country_colors to df based on matching country names
df = df.merge(df_country_colors, how='left', left_on='team_country', right_on='Country')
df['age_years'] = df['age'].str.extract(r'(\d+)').astype(float)
df['position_first'] = df['position'].str.split(',').str[0]
df['pens_missed'] = df['pens_att'] - df['pens_made']
df.head(5)
```

<b>Make a Scatter Plot</b>
Next, we'll take a look at some of the data relationships we have at the player level by making a scatter plot. Feel free to try out different combinations of fields to see if there any interesting or noteworthy data relationships.

```python
scatter_field_sum_x = 'age_years'
scatter_field_sum_y = 'goals_assists'
scatter_global_font = 'Arial'
scatter_title_font_size = 24
scatter_title = f"Scatter Plot of {scatter_field_sum_y} vs {scatter_field_sum_x} for Players in the 2026 World Cup"
```

```python
# Scatter plot using plotly that shows the relationship between player age_years and goals_assists. Color should be by Primary Color HEX. Hover should show player name, team_country, age_years, club, position_first, and goals_assists. Only include players with one or more goals_assists
df_filtered_scatter = df[df[scatter_field_sum_y] > 0]

fig = go.Figure()
fig.add_trace(
    go.Scatter(
    x=df_filtered_scatter[scatter_field_sum_x],
    y=df_filtered_scatter[scatter_field_sum_y],
    mode='markers',
    marker=dict(
        color=df_filtered_scatter['Primary Color HEX'],  # Color by Primary Color HEX
        size=15,
        line=dict(width=1, color='DarkSlateGrey')
    ),
    text=df_filtered_scatter.apply(lambda row: f"Player: {row['player']}<br>Country: {row['team_country']}<br>Club: {row['club']}<br>Position: {row['position_first']}<br>{scatter_field_sum_y}: {row[scatter_field_sum_y]}<br>{scatter_field_sum_x}: {row[scatter_field_sum_x]}", axis=1),
    hoverinfo='text'
)
)
# Display the chart
fig.update_layout(
    xaxis_title=scatter_field_sum_x.replace('_', ' ').title(),
    yaxis_title=scatter_field_sum_y.replace('_', ' ').title(),
    font=dict(
        family=scatter_global_font
,
        size=14,
        color="Black"
    ),
    title=dict(
            text=f"<b>{scatter_title}</b>",
            font=dict(
                family=scatter_global_font,
                size=scatter_title_font_size,
                color="Black"
            )
        ),
    template='plotly_white'
)
fig.show()
```

<b>Make a stacked bar chart</b><br>
The following is a look at penalties made and missed by country. It's also an example of a dark mode themed graph. We can play around with aesthetics here to see what has the most impact and what we like.

```python
# User variables for penalties graph
penalties_outer_background = '#010b1c'
penalties_inner_background = '#01122e'
penalties_missed_bar_color = '#e33b3b'
penalaties_made_bar_color = '#66c46e'
penalties_global_font = 'Arial'
penalties_title_font_size = 24
penalties_font_color = 'white'
```

```python
# Stacked bar chart from df that shows pens_made vs pens_attempted by team_country. Only include countries with one or more pens_attempted. Color for pens_made should be green and color for pens_missed should be red. Hover should show team_country, pens_made, pens_missed, and pens_attempted.

df_pens_filtered_stacked = df[df['pens_att'] > 0].groupby('team_country', as_index=False).agg({
    'pens_made': 'sum',
    'pens_missed': 'sum',
    'pens_att': 'sum'
})

fig = go.Figure()
fig.add_trace(
    go.Bar(
        x=df_pens_filtered_stacked['team_country'],
        y=df_pens_filtered_stacked['pens_made'],
        name='Pens Made',
        marker_color=penalaties_made_bar_color
    )
)
fig.add_trace(
    go.Bar(
        x=df_pens_filtered_stacked['team_country'],
        y=df_pens_filtered_stacked['pens_missed'],
        name='Pens Missed',
        marker_color=penalties_missed_bar_color
    )
)
fig.update_layout(
    barmode='stack',
    xaxis=dict(
        title=f"<b>Countries</b>",
        showgrid=False
        ),
    yaxis=dict(
        title=f"<b>Pens Attempted</b>",
        side="left",
        showgrid=False,
        tickformat=',d', # Format y-axis ticks as integers
        dtick=1
        ),
    font=dict(
        family=penalties_global_font,
        size=14,
        color=penalties_font_color
    ),
    title=dict(
            text=f"<b>Made and Missed Penalties by Country</b>",
            font=dict(
                family=penalties_global_font,
                size=penalties_title_font_size,
                color=penalties_font_color
            )
        ),
    paper_bgcolor= penalties_outer_background,
    plot_bgcolor=penalties_inner_background,
    template='plotly_white'
)
fig.show()
```

<b>Next we'll be making graphs that are a bit more complex: Pareto charts</b><br>
Pareto charts focus on showing the primary inputs or contributors on a result. There especially helpful when there are MANY contributors. Even more so when a small group of those contributors have an outsized impact on the result. These charts are tied to the 80/20 rule which states: in a lot of cases, 80% of your outputs (results) are driven by 20% of your inputs (contributors). Note, it doesn't have to be an 80/20 split for a Pareto to still be the right choice for a graph. <br><br>
Here, we'll look at the impact of countries and then individual players on overall goal output for the 2026 World Cup.

```python
df_goals_check = df.sort_values(by='goals', ascending=False)
df_goals_check.head(5)
```

```python
# Sort df so you can calculate running some
df_pareto_players = df.sort_values(by='goals', ascending=False).reset_index(drop=True)

# Calculate the cumulative sum of goals
df_pareto_players['cumulative_goals'] = df_pareto_players['goals'].cumsum()
df_pareto_players['cumulative_percentage'] = df_pareto_players['cumulative_goals'] / df_pareto_players['goals'].sum() * 100
```

```python
df_pareto_players.head(5)
```

```python
# Pareto switch variables
## You set
pareto_threshold = 20  # 80% of goals
## Calculated
pareto_player_count = df_pareto_players[df_pareto_players['cumulative_percentage'] < pareto_threshold].shape[0]
pareto_other_sum = df_pareto_players[df_pareto_players['cumulative_percentage'] >= pareto_threshold]['goals'].sum()
pareto_other_percentage = pareto_other_sum / df_pareto_players['goals'].sum() * 100
pareto_other_player_count = df_pareto_players[df_pareto_players['cumulative_percentage'] >= pareto_threshold].shape[0]
pareto_total_goals = df_pareto_players['goals'].sum()

pareto_title = f"Pareto Analysis: {pareto_player_count} Players Account for {pareto_threshold}% of the {pareto_total_goals:.0f} Total Goals Scored at the 2026 World Cup"
```

```python
# Create list of df_pareto players that are above the pareto threshold
df_pareto_players_above_threshold = df_pareto_players[df_pareto_players['cumulative_percentage'] < pareto_threshold]
# Create list of player names from df_pareto players that are above the pareto threshold
pareto_players_above_threshold = df_pareto_players_above_threshold['player'].tolist()
# Create list of goal values from df_pareto players that are above the pareto threshold
pareto_goals_above_threshold = df_pareto_players_above_threshold['goals'].tolist()
# Create list of cumulative goal percentages from df_pareto players that are above the pareto threshold
pareto_cumulative_goals_above_threshold = df_pareto_players_above_threshold['cumulative_percentage'].tolist()
pareto_percentage_labels = [f"{percentage:.1f}%" for percentage in pareto_cumulative_goals_above_threshold]
```

```python
# Pareto aesthetic variables
pareto_player_graph_outer_background = 'white'
pareto_player_graph_inner_background = 'white'
pareto_player_line_color = "crimson"
pareto_player_bar_color = "#029c75"
pareto_player_bar_opacity = 0.5
pareto_player_show_legend = False
pareto_player_global_font = "Arial"
pareoto_player_title_font_size = 24
```

```python
# Make pareto chart with plotly
fig = go.Figure()

#Add the primary Bar Chart trace (Left Y-Axis)
fig.add_trace(
    go.Bar(
        x=pareto_players_above_threshold,
        y=pareto_goals_above_threshold,
        name="Player",
        marker_color=pareto_player_bar_color,
        opacity=pareto_player_bar_opacity,
        yaxis='y1',
        customdata=df_pareto_players_above_threshold[['team_country']].values,  # Add country data for hover
        hovertemplate=(
            '<b>%{x}</b><br>'+
            'Goals: %{y}<br>'+
            'Country: %{customdata[0]}<br>'+
            '<extra></extra>')
    )
)

# 6. Add the secondary Line Chart trace (Right Y-Axis)
fig.add_trace(
    go.Scatter(
        x=df_pareto_players_above_threshold['player'],
        y=pareto_cumulative_goals_above_threshold,
        name="Cumulative %",
        mode="lines+markers+text",
        text=pareto_percentage_labels,
        textposition='top center',
        textfont=dict(
            color='black', 
            size=14
        ),
        line=dict(
            color=pareto_player_line_color, 
            width=2, 
            dash='dot',
            shape='spline'
        ),
        marker=dict(
            color=pareto_player_line_color, 
            size=16,
            symbol='circle'
        ),
        yaxis="y2"  # Points this trace to the secondary y-axis
    )
)

# 7. Configure layout with dual axes
fig.update_layout(
    showlegend=pareto_player_show_legend,
    xaxis=dict(
        title="Players",
        showgrid=False
        ),
    yaxis=dict(
        title="Goals",
        side="left",
        showgrid=False
        ),
    yaxis2=dict(
        title=f"Cumulative % of all Goals",
        side="right",
        overlaying="y",  # Ensures it overlays the primary y-axis
        range=[0, 105],  # Pads the top slightly for visual clarity
        ticksuffix="%",  # Adds percent signs to right-hand axis labels
        showgrid=False
    ),
    font=dict(
        family=pareto_player_global_font,
        size=14,
        color="Black"
    ),
    title=dict(
            text=f"<b>{pareto_title}</b>",
            font=dict(
                family=pareto_player_global_font,
                size=pareoto_player_title_font_size,
                color="Black"
            )
        ),
    legend=dict(x=0.8, y=1.1), # Positions the legend neatly
    paper_bgcolor= pareto_player_graph_outer_background,
    plot_bgcolor=pareto_player_graph_inner_background,
    template="plotly_white",
    )
# Display the interactive chart
fig.show()
```

<b>Now we'll make a country based Pareto!</b>

```python
# Create a dataframe from df of countries and total goals scored by players from that country
df_country_goals = df.groupby(['team_country', 'Primary Color HEX', 'Secondary Color HEX'], as_index=False).agg({'goals': 'sum'}).rename(columns={'goals': 'total_goals'})

# Sort df_country_goals by total_goals in descending order
df_pareto_country = df_country_goals.sort_values(by='total_goals', ascending=False)
# Add cumulative goals and cumulative percentage columns to df_pareto_country
df_pareto_country['cumulative_goals'] = df_pareto_country['total_goals'].cumsum()
df_pareto_country['cumulative_percentage'] = df_pareto_country['cumulative_goals'] / df_pareto_country['total_goals'].sum() * 100
```

```python
# Country Pareto switch variables
## You set
country_pareto_threshold = 40 # % of total goals
## Calculated
country_pareto_player_count = df_pareto_country[df_pareto_country['cumulative_percentage'] < country_pareto_threshold].shape[0]
country_pareto_other_sum = df_pareto_country[df_pareto_country['cumulative_percentage'] >= country_pareto_threshold]['total_goals'].sum()
country_pareto_other_percentage = country_pareto_other_sum / df_pareto_country['total_goals'].sum() * 100
country_pareto_other_player_count = df_pareto_country[df_pareto_country['cumulative_percentage'] >= country_pareto_threshold].shape[0]
country_pareto_total_goals = df_pareto_country['total_goals'].sum()

country_pareto_title = f"Pareto Analysis: {country_pareto_player_count} Countries Account for {country_pareto_threshold}% of the {country_pareto_total_goals:.0f} Total Goals Scored at the 2026 World Cup"
```

```python
# Create list of df_pareto countries that are above the pareto threshold
df_pareto_countries_above_threshold = df_pareto_country[df_pareto_country['cumulative_percentage'] < country_pareto_threshold]
```

```python
# Pareto aesthetic variables
pareto_country_graph_outer_background = 'white'
pareto_country_graph_inner_background = 'white'
pareto_country_line_color = "crimson"
pareto_country_bar_opacity = 0.5
pareto_country_show_legend = False
pareto_country_global_font = "Arial"
pareto_country_title_font_size = 24
```

```python
# Make pareto chart with plotly
fig = go.Figure()

#Add the primary Bar Chart trace (Left Y-Axis)
fig.add_trace(
    go.Bar(
        x=df_pareto_countries_above_threshold['team_country'],
        y=df_pareto_countries_above_threshold['total_goals'],
        name="Player",
        opacity=pareto_player_bar_opacity,
        marker=dict(
            color=df_pareto_countries_above_threshold['Primary Color HEX'], # Fills the inner bar area
            line=dict(
                color=df_pareto_countries_above_threshold['Secondary Color HEX'], # Sets the border color
                width=4 # Sets the border line width in pixels
            )
        ),
        yaxis='y1',
        customdata=df_pareto_countries_above_threshold[['cumulative_goals']].values,  # Add country data for hover
        hovertemplate=(
            '<b>%{x}</b><br>'+
            'Team Goals: %{y}<br>'+
            'Cumulative Goals: %{customdata[0]}<br>'+
            '<extra></extra>')
    )
)

# 6. Add the secondary Line Chart trace (Right Y-Axis)
fig.add_trace(
    go.Scatter(
        x=df_pareto_countries_above_threshold['team_country'],
        y=df_pareto_countries_above_threshold['cumulative_percentage'],
        name="Cumulative %",
        mode="lines+markers+text",
        text=df_pareto_countries_above_threshold['cumulative_percentage'].apply(lambda x: f"{x:.1f}%"),  # Format as percentage 
        textposition='top center',
        textfont=dict(
            color='black', 
            size=14
        ),
        line=dict(
            color=pareto_country_line_color, 
            width=2, 
            dash='dot',
            shape='spline'
        ),
        marker=dict(
            color=pareto_country_line_color, 
            size=16,
            symbol='circle'
        ),
        yaxis="y2"  # Points this trace to the secondary y-axis
    )
)

# 7. Configure layout with dual axes
fig.update_layout(
    showlegend=pareto_country_show_legend,
    xaxis=dict(
        title="Countries",
        showgrid=False
        ),
    yaxis=dict(
        title="Goals",
        side="left",
        showgrid=False
        ),
    yaxis2=dict(
        title=f"Cumulative % of all Goals",
        side="right",
        overlaying="y",  # Ensures it overlays the primary y-axis
        range=[0, 105],  # Pads the top slightly for visual clarity
        ticksuffix="%",  # Adds percent signs to right-hand axis labels
        showgrid=False
    ),
    font=dict(
        family=pareto_country_global_font,
        size=14,
        color="Black"
    ),
    title=dict(
            text=f"<b>{country_pareto_title}</b>",
            font=dict(
                family=pareto_country_global_font,
                size=pareto_country_title_font_size,
                color="Black"
            )
        ),
    legend=dict(x=0.8, y=1.1), # Positions the legend neatly
    paper_bgcolor= pareto_country_graph_outer_background,
    plot_bgcolor=pareto_country_graph_inner_background,
    template="plotly_white",
    )
# Display the interactive chart
fig.show()
```
